```python
def relative_unc(sig_eff, sig_eff_true, ref_eff, ref_eff_true):
    # d_epsilon_sig = abs(sig_eff - sig_eff_true)
    # d_epsilon_ref = abs(ref_eff - ref_eff_true)
    d_epsilon_sig = sig_eff - sig_eff_true
    d_epsilon_ref = ref_eff - ref_eff_true
    rel_unc = d_epsilon_sig/sig_eff - d_epsilon_ref/ref_eff
    abs_rel_unc = abs(rel_unc)
    print(f"d_epsilon_sig/sig_eff = {d_epsilon_sig/sig_eff*100:.6f}%")
    print(f"d_epsilon_ref/ref_eff = {d_epsilon_ref/ref_eff*100:.6f}%")

    print(f"rel_unc = {abs_rel_unc*100:.6f}%")
    # print(f"rel_unc = {abs_rel_unc:.6f}")
```
