```python
Br_kpi = 0.039500000
Br_kpipipi = 0.008800000
Br_kk = 0.004080000
```

```python
import math

# Given values
y = 1795
dy = math.sqrt(y)
x = 43542
dx = math.sqrt(x)

f = y / x
# Relative error propagation
rel_error = math.sqrt((dy / y) ** 2 + (dx / x) ** 2)
# Absolute uncertainty
df = f * rel_error

difference = f - Br_kpi
sigma = difference / df

print(f"f = {f:.6f} ± {df:.6f}")
print(f"{sigma:.6f} sigma difference")
```
