```python
def calculate_sig_eff_err(eff, N_gen):

    error = math.sqrt(eff * (1 - eff) / N_gen)
    return error
```

```python
eff_input = 0.54257
N_gen_input = 1e+6

error = calculate_sig_eff_err(eff_input, N_gen_input)
print(f"eff: {eff_input:.4f} \pm {error:.4f}")
```
