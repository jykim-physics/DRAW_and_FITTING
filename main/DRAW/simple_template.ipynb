```python
import numpy as np
import matplotlib.pyplot as plt

def func(x, S):
    return np.sqrt((1 - x**2) / S)

x = np.linspace(0, 0.001, 1000)
# a_values = [34643, 50000, 77052, 8*10**5,1000000 ]
a_values = [120000, 34000,476000, 4560000]

plt.figure(figsize=(8, 6))
for S in a_values:
    y = func(x, S)
    plt.plot(x, y, label=f'S = {S}')

plt.xlabel(r'$A_{CP}$')
plt.ylabel(r'$\sigma_{A_{CP}}$')
plt.title(r'$\sigma_{A_{CP}} = \sqrt{(1 - {A_{CP}}^2) / S}$')
plt.xlim(0,)
plt.ylim(0,)
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define ranges for S and B
S_values = np.linspace(1, 100, 100)  # Signal from 1 to 100
B_small = 5   # Small background
B_large = 50  # Large background

# Compute FOMs
FOM_standard_small_B = S_values / np.sqrt(S_values + B_small)
FOM_modified_small_B = S_values**(3/2) / np.sqrt(S_values + B_small)

FOM_standard_large_B = S_values / np.sqrt(S_values + B_large)
FOM_modified_large_B = S_values**(3/2) / np.sqrt(S_values + B_large)

# Plot
plt.figure(figsize=(10, 6))

plt.plot(S_values, FOM_standard_small_B, label=r'Standard FOM ($B=5$)', linestyle='--', color='blue')
plt.plot(S_values, FOM_modified_small_B, label=r'Modified FOM ($B=5$)', linestyle='-', color='blue')

plt.plot(S_values, FOM_standard_large_B, label=r'Standard FOM ($B=50$)', linestyle='--', color='red')
plt.plot(S_values, FOM_modified_large_B, label=r'Modified FOM ($B=50$)', linestyle='-', color='red')

plt.xlabel("Signal Events (S)")
plt.ylabel("Figure of Merit")
plt.title("Comparison of Standard and Modified FOMs")
plt.legend()
plt.grid(True)

plt.show()
```
